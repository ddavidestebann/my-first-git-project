# Evolucion del codigo aplicando principios de SDLC

Cada bloque mejora el bloque anterior. El ejemplo trabaja con una lista de numeros aleatorios y calcula su promedio y maximo.

## Punto de partida: codigo raw / messy

Este codigo funciona para un caso puntual, pero mezcla generacion de datos, calculos y salida. No se puede reutilizar facilmente y falla si la lista esta vacia.

```python
import random

numbers = [random.randint(1, 100) for _ in range(10)]
print("Generated numbers:", numbers)

total = 0
for number in numbers:
    total += number
average = total / len(numbers)
print("Average:", average)

max_number = numbers[0]
for number in numbers:
    if number > max_number:
        max_number = number
print("Max:", max_number)
```

## Task 1: Modularity

Separamos cada responsabilidad en una funcion. Ahora cada parte puede probarse y cambiarse de manera independiente.

```python
import random

def generate_numbers():
    return [random.randint(1, 100) for _ in range(10)]

def calculate_average(numbers):
    return sum(numbers) / len(numbers)

def find_max(numbers):
    return max(numbers)

numbers = generate_numbers()
print("Generated numbers:", numbers)
print("Average:", calculate_average(numbers))
print("Max:", find_max(numbers))
```

## Task 2: Reusability

Las funciones reciben parametros en lugar de depender de valores fijos. Asi pueden usarse con otros tamanos, rangos y listas en cualquier proyecto.

```python
from typing import List

def generate_numbers(count: int, lower: int = 1, upper: int = 100) -> List[int]:
    return [random.randint(lower, upper) for _ in range(count)]

def calculate_average(numbers: List[int]) -> float:
    return sum(numbers) / len(numbers)

def find_max(numbers: List[int]) -> int:
    return max(numbers)

small_sample = generate_numbers(count=5, lower=10, upper=20)
large_range_sample = generate_numbers(count=5, lower=1000, upper=2000)

print("Sample 1:", small_sample)
print("Average 1:", calculate_average(small_sample))
print("Max 2:", find_max(large_range_sample))
```

## Task 3: Maintainability

Centralizamos los calculos en `summarize_numbers`. Agregar minimo, mediana u otra metrica requiere extender un solo lugar, sin repetir el flujo principal.

```python
from statistics import median

def summarize_numbers(numbers: List[int]) -> dict:
    return {
        "average": calculate_average(numbers),
        "maximum": find_max(numbers),
        "minimum": min(numbers),
        "median": median(numbers),
    }

numbers = generate_numbers(count=10)
print("Numbers:", numbers)
print("Summary:", summarize_numbers(numbers))
```

## Task 4: Scalability

El algoritmo no cambia al aumentar la cantidad de datos. Solo modificamos `count`; las mismas funciones procesan una lista mucho mayor.

```python
numbers = generate_numbers(count=100_000, lower=1, upper=1_000_000)
summary = summarize_numbers(numbers)

print("Number of values:", len(numbers))
print("Scalable summary:", summary)
```

## Task 5: Reliability & Quality

Agregamos validaciones y errores claros. Esto evita resultados silenciosamente incorrectos cuando los parametros no son validos.

```python
def generate_numbers(count: int, lower: int = 1, upper: int = 100) -> List[int]:
    if not isinstance(count, int) or isinstance(count, bool):
        raise TypeError("count must be an integer")
    if count < 0:
        raise ValueError("count cannot be negative")
    if lower > upper:
        raise ValueError("lower must be less than or equal to upper")
    return [random.randint(lower, upper) for _ in range(count)]

try:
    generate_numbers(count=-1)
except (TypeError, ValueError) as error:
    print("Validation error captured:", error)
```

## Task 6: Security & Trust

Validamos explicitamente la entrada vacia antes de calcular estadisticas. La funcion no usa `eval`, archivos ni entradas ejecutables: solo acepta datos y parametros controlados.

```python
def require_numbers(numbers: List[int]) -> None:
    if not numbers:
        raise ValueError("The list of numbers cannot be empty")

def calculate_average(numbers: List[int]) -> float:
    require_numbers(numbers)
    return sum(numbers) / len(numbers)

def find_max(numbers: List[int]) -> int:
    require_numbers(numbers)
    return max(numbers)

def summarize_numbers(numbers: List[int]) -> dict:
    require_numbers(numbers)
    return {
        "average": calculate_average(numbers),
        "maximum": find_max(numbers),
        "minimum": min(numbers),
        "median": median(numbers),
    }

try:
    summarize_numbers([])
except ValueError as error:
    print("Empty input rejected:", error)

trusted_numbers = generate_numbers(count=10)
print("Trusted summary:", summarize_numbers(trusted_numbers))
```

## Resultado

El codigo final es modular, reutilizable, mantenible, escalable y valida sus entradas. Cada mejora conserva lo aprendido en el bloque anterior.
